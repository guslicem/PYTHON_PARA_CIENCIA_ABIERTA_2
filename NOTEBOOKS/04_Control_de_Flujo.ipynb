# 1. Operadores lógicos y relacionales en Python

En Python podemos usar **operadores relacionales** (comparaciones) y **operadores lógicos** (combinación de condiciones).  
Ambos son fundamentales para tomar decisiones en los programas.

---

## Tabla de operadores

| Tipo           | Operador | Descripción | Ejemplo | Resultado |
|----------------|----------|-------------|---------|-----------|
| Relacional     | `==`     | Igual a     | `5 == 5` | `True` |
| Relacional     | `!=`     | Distinto de | `5 != 3` | `True` |
| Relacional     | `>`      | Mayor que   | `7 > 3`  | `True` |
| Relacional     | `<`      | Menor que   | `2 < 8`  | `True` |
| Relacional     | `>=`     | Mayor o igual que | `5 >= 5` | `True` |
| Relacional     | `<=`     | Menor o igual que | `4 <= 2` | `False` |
| Lógico         | `and`  `&`  | Devuelve `True` si **todas** las condiciones son verdaderas | `(5 > 2 and 3 < 4)` | `True` |
| Lógico         | `or`    `\|` | Devuelve `True` si **alguna** condición es verdadera | `(5 < 2 or 3 < 4)` | `True` |
| Lógico         | `not`    | Invierte el valor lógico | `not(5 > 2)` | `False` |

---


`````python

## Ejemplos

a = 10
b = 5

# Operadores relacionales
print(a == b)   # False
print(a != b)   # True
print(a > b)    # True
print(a <= b)   # False

# Operadores lógicos
print(a > 0 and b > 0)   # True, ambas condiciones son ciertas
print(a < 0 or b > 0)    # True, la segunda condición es cierta
print(not(a > b))        # False, porque a > b es True y NOT lo invierte
`````

# 2. Condicionales en Python: if, elif, else

En Python podemos usar estructuras condicionales para **ejecutar instrucciones solo cuando se cumpla una condición**.

---

## Sintaxis básica: if ... else

La forma más sencilla de condicional es:

```python

    # bloque de código si la condición es True

`````python
edad = 20

if edad >= 18:
    print("Te puedes pedir una cerveza")
else:
    print("Para ti, mejor un Colacao fesquito")
`````

# Uso de `elif` en Python

Cuando tenemos **más de dos condiciones posibles**, no basta con `if ... else`.  
En esos casos usamos `elif` (*else if*), que nos permite encadenar varias condiciones de forma clara y ordenada.

```python

elif condicion2:
    # bloque de código si condicion1 es False y condicion2 es True
elif condicion3:

    # bloque de código si ninguna condición es True

# 3. Uso de `match-case` en Python

En Python 3.10 y versiones posteriores, podemos usar `match-case` como alternativa a largas cadenas de `if-elif-else`.  
`match-case` permite **comparar un valor con diferentes patrones** y ejecutar el bloque correspondiente.

No es algo que se use mucho, a mi criterio, es más habitual encontrar if--elif--else
---

## Sintaxis básica

```python
match variable:
    case valor1:
        # bloque de código si variable == valor1
    case valor2:
        # bloque de código si variable == valor2
    case _:
        # bloque de código si ningún valor coincide (equivalente a else)


`````python
#Ejemplo

dia = "sábado"

match dia:
    case "lunes":
        print("Hoy es el peor día de la semana")
    case "sábado" | "domingo":
        print("Estás de finde!!!")
    case _:
        print("Día entre semana")

#juega a cambiar el dia
`````

# 3. El bucle `for` en Python

El bucle `for` en Python se utiliza para **recorrer elementos de una secuencia** (listas, cadenas de texto, rangos de números, diccionarios, etc.) y ejecutar un bloque de instrucciones repetidamente.

---

## Sintaxis básica
```python
for variable in secuencia:
    # bloque de instrucciones
````
variable toma en cada iteración el valor de un elemento de la secuencia.

El bucle se repite hasta recorrer todos los elementos, por orden

## Instrucciones de control en los bucles: `break` y `continue`

En Python podemos alterar el flujo normal de un bucle usando estas dos instrucciones:

### `break`
- Sirve para **interrumpir el bucle por completo** y salir de él inmediatamente.
- Se usa, por ejemplo, cuando ya hemos encontrado lo que buscábamos y no tiene sentido seguir iterando.


`````python

#### Ejemplo:
secuencia = [10,11,12,5, 13,14]

for numero in secuencia:
    print(numero)

#### Ejemplo de destruccion masiva!!!
secuencia = [10,11,12,5, 13,14]

for numero in secuencia:
    secuencia.append(numero+1)
    print(numero)
    if(len(secuencia)>1000):
        break
"""
El bucle for en Python no copia la lista, sino que la recorre elemento a elemento.

-- En cada iteración mira el siguiente índice de la lista dinámicamente.

-- Esto significa que si la lista crece, el bucle seguirá recorriendo los nuevos elementos añadidos.

Como dentro del for se hace secuencia.append(numero+1), la lista crece indefinidamente

sin el if..... seguiría hasta colapsar la memoria del equipo ya que la variable secuencia 

crecería de tamaño constantemente
"""
`````

La función `range()` genera secuencias de números enteros, muy útil para controlar bucles.

`````python
# range(inicio, fin, paso)
# inicio: primer número (por defecto 0)
# fin: se detiene en fin-1
# paso: incremento (por defecto 1)

for i in range(5):
    print(i)  # 0, 1, 2, 3, 4 Recuerda, siempre el primer elemento está en la direccion 0
print('------------')

Valores= range(10)
iteracion = 1
print(type(Valores))
for elemento in Valores:
    print(f'valor: {elemento} Iteracion: {iteracion}')
    iteracion += 1 
    #Aprendemos este modo de incrementar/decrementar una variable
    # a += 1 es lo mismo que a = a+1
    
print('------------')

for i in range(2, 10, 2):
    print(i)  # 2, 4, 6, 8
`````

`````python

for numero in range(5):
    if numero == 2:
        continue  # saltamos la iteración cuando numero == 2
    print(numero)
`````


### `continue`

Sirve para saltar a la siguiente iteración sin ejecutar el resto del código dentro del bucle en esa vuelta.

Se usa, por ejemplo, para ignorar ciertos valores pero continuar con el bucle.


A veces queremos detener el bucle antes de tiempo. Para ello se usa break. Por ejemplo lo ejecutamos mientras una variable, cuya evolucion desconocemos, cumpla alguna condicion. Imagina que tienes un csv de lecturas de temperatura que van creciendo, junto a otras lecturas de, pongamos radiacion UV, y quieres ir reportando la radiación UV mientras la temperatura fue menor de 40ºC. Para ello tenemos break.

`````python
for i in range(10):
    print(i)

    if i == 5:
        print("Me paro en 5")
        break
`````

Si queremos recorrer en un mismo bucle varias secuencias tenemos la sentencia zip()
Uso de `zip()` en bucles `for`

La función `zip()` nos permite **recorrer varias secuencias a la vez**.  
Combina elementos de varias listas (o secuencias) en tuplas, emparejando los elementos que ocupan la misma posición.

---

## Sintaxis
```python
for a, b, c, ... in zip(seq1, seq2, seq3, ...):
    # bloque de instrucciones

```
Cada iteración devuelve una `tupla` con los elementos de las secuencias en la misma posición.

**IMPORTANTE** El bucle se detiene cuando se alcanza el final de la secuencia más corta.

`````python
nombres = ["Ana", "Luis", "Marta"]
edades = [25, 30, 22,35]
ciudades = ["Madrid", "Sevilla", "Valencia","Caceres", "Barcelona"]
for nombre, edad, ciudad in zip(nombres, edades, ciudades):
    print(f"{nombre} tiene {edad} años y vive en {ciudad}")
`````

# 4. Bucle `while` en Python

El bucle `while` repite un bloque de instrucciones **mientras** una condición sea verdadera.  
A diferencia del `for`, no recorre de manera explícita una lista o un rango, sino que depende de una **condición lógica** que se evalúa en cada iteración.

La sintaxis básica es:

```python
while condición:
    # bloque de código
```

Un error **MUY HABITUAL** es no inicializar la variable que pondremos en la condición, o no actualizarla dentro del while lo que nos llevará a bucles infinitos 

# Diferencias clave entre `for` y `while`

- El bucle `for` recorre una secuencia (lista, rango, cadena, diccionario, etc.).  
  Si la secuencia tiene al menos un elemento, **el bucle siempre se ejecuta al menos una vez**.  

- El bucle `while` depende únicamente de que la condición inicial sea verdadera.  
  Si la condición es falsa desde el principio, **el bucle puede no ejecutarse nunca**.  



`````python
#Bucle básico


i = 0
while i < 5:
    print(i)
    i += 1 # i = i+1

`````

`````python
contador = 0
while contador < 10:
    contador += 1
    if contador % 2 == 0: #El resto en modulo 2 igual a cero implica numero par
        continue  # salta el resto del código y vuelve al inicio del bucle
    print("Número:", contador)

`````

### Bucle while con entrada del usuario

`````python
#En este ejemplo nos quedamos dentro del bucle hasta que no escribamos salir

respuesta = "" #Inicializamos la variable de respuesta a una cadena vacia.

while respuesta != "salir": 
    respuesta = input("Escribe algo (o 'salir' para terminar): ")
    #Ahora vamos a evitar que el usuario se nos ponga creativo
    respuesta = respuesta.lower() #toda entera en minusculas
    
    print("Has escrito:", respuesta)
`````

`````python
#Añadimos un break para que el bucle mas no se ejecute si ha llegado a la iteracion 5

contador = 0
while True:
    contador += 1
    print("Iteración:", contador)
    if contador == 5:
        break
`````

# Contenido Avanzado: Manejo de Errores con try-except

En Python, algunas entradas de usuario o operaciones pueden provocar errores que detienen el programa.  
Para controlar esto y evitar que el programa se rompa, usamos **try-except**.

---

## Sintaxis básica

```python
try:
    # Código que puede generar un error
    numero = int(input("Introduce un número: "))
    print("El doble es:", numero * 2)

except ValueError:
    # Este bloque se ejecuta si ocurre un ValueError
    print("¡Eso no es un número válido!")


`````python
#Vamos a forzar un error

numero = "A"
print(f"El cuadrado de {numero} es {numero**2}")
`````

`````python

try:
    # Código que puede generar un error
    numero = int(input("Introduce un número: "))
    print(f"El cuadrado de {numero} es {numero**2}")

except ValueError:
    # Este bloque se ejecuta si ocurre un ValueError
    print("¡Eso no es un número válido!")
`````

`````python
"""
Contenido avanzado: pedir un número válido usando try-except y while.

En este ejemplo combinamos try-except con while para que el programa
siga pidiendo un número hasta que el usuario introduzca uno válido.
"""


numero_valido = False  # Un bool que nos dice si hemos o no introducido un numero

while not numero_valido:
    try:
        numero = int(input("Introduce un número: "))
        print(f"El cuadrado de {numero} es {numero**2}")
        numero_valido = True  # Si llegamos aquí, la entrada es válida

    except ValueError:
        print("¡Eso no es un número válido! Inténtalo de nuevo.")
`````

`````python
# SOLUCIONES

# Ejercicio 1: Número par o impar
print("Ejercicio 1")
numero = int(input("Introduce un número: "))
if numero % 2 == 0:
    print(f"{numero} es par")
else:
    print(f"{numero} es impar")

# Ejercicio 2
print("Ejercicio 2")

a = float(input("Introduce el primer número: "))
b = float(input("Introduce el segundo número: "))

if a > b:
    print(f"{a} es mayor que {b}")
elif a < b:
    print(f"{b} es mayor que {a}")
else:
    print("Los números son iguales")

# Ejercicio 3
print("Ejercicio 3")

edad = int(input("Introduce tu edad: "))

if edad < 13:
    print("Niño")
elif edad <= 19:
    print("Adolescente")
elif edad <= 59:
    print("Adulto")
else:
    print("Adulto mayor")

# Ejercicio 4
print("Ejercicio 4")

#Solucion básica

frase = input("Introduce una frase: ")

# Inicializamos los contadores
mayusculas = 0
minusculas = 0

# Recorremos cada carácter de la frase
for c in frase:
    if c.isupper():      # Si es una letra mayúscula
        mayusculas += 1  # Incrementamos el contador de mayúsculas
    elif c.islower():    # Si es una letra minúscula
        minusculas += 1  # Incrementamos el contador de minúsculas

# Mostramos los resultados
print("Número de letras mayúsculas:", mayusculas)
print("Número de letras minúsculas:", minusculas)

#Solucion más avanzada

mayusculas = sum(1 for c in frase if c.isupper()) #sumo 1 por cada caracter en la cadena que sea una mayúscula
minusculas = sum(1 for c in frase if c.islower()) #sumo 1 por cada caracter en la cadena que sea una minúscula

print(f"Mayúsculas: {mayusculas}, Minúsculas: {minusculas}")

# Ejercicio 5
print("Ejercicio 5")

n = int(input("Introduce un número entero positivo: "))
factorial = 1

for i in range(1, n + 1):
    factorial *= i

print(f"El factorial de {n} es {factorial}")


# Ejercicio 6
print("Ejercicio 6")


numeros = [3, 8, 12, 5, 7, 10]

for num in numeros:
    if num > 6:
        print(num)

# Ejercicio 7
print("Ejercicio 7")

nombres = [input("Nombre 1: "), input("Nombre 2: "), input("Nombre 3: ")]

edades = [int(input(f"Edad de {nombre}: ")) for nombre in nombres]

diccionario = dict(zip(nombres, edades))

for nombre, edad in diccionario.items():
    print(f"{nombre} tiene {edad} años")

# Ejercicio 8
print("Ejercicio 8")

suma = 0

while suma <= 50:
    numero = float(input("Introduce un número: "))
    suma += numero
    print(f"Suma actual: {suma}")

print(f"Suma final: {suma}")

# Ejercicio 9
print("Ejercicio 9")

texto = input("Introduce una palabra o frase: ").replace(" ", "").lower()
if texto == texto[::-1]:
    print("Es un palíndromo")
else:
    print("No es un palíndromo")

# Ejercicio 10
print("Ejercicio 10")

numero_valido = False

while True:
    try:
        numero = int(input("Introduce un número, no pararé de preguntar hasta que me des un impar: "))
        if numero % 2 != 0:
            print(f"¡Número impar {numero} recibido! Fin del bucle.")
            break
    except ValueError:
        print("Eso no es un número válido, inténtalo de nuevo.")

`````
