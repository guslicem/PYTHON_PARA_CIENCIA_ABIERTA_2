## 5. Paso 3: Instalación del Paquete en Modo Editable (`pip install -e .`)

Durante la fase de investigación y desarrollo, es muy conveniente instalar el paquete en **modo editable** (`-e` o `--editable`).

### ¿Qué hace el flag `-e`?
En lugar de copiar los archivos del paquete al directorio de librerías de Python (`site-packages`), la instalación en modo editable crea una **referencia simbólica o enlace** hacia tu carpeta de código en desarrollo.

Esto significa que **cualquier cambio que guardes en los archivos `.py` de la carpeta `src/` se reflejará inmediatamente en tus cuadernos sin necesidad de volver a ejecutar `pip install`**.

```bash
# Navegamos a la carpeta del paquete
cd ./MI_PAQUETE_CSIC

# Instalamos el paquete localmente en modo editable
pip install -e .
```


## 6. Desarrollo de Módulos: Funciones Creadas y Clases de Dominio Científico

A continuación implementamos en el código fuente de nuestro paquete dos componentes fundamentales:
1. Una función matemática recursiva (`my_factorial`).
2. Una clase de dominio para gestionar investigadores y personal de apoyo del CSIC (`Persona`).


````python
# 1. Definición e Implementación de la Función Factorial en el Módulo de Ejemplo
def my_factorial(x: int) -> int:
    '''
    Calcula el factorial de un número entero no negativo utilizando recursión.

    Parameters
    ----------
    x : int
        Número entero no negativo.

    Returns
    -------
    int
        El factorial de x.

    Raises
    ------
    ValueError
        Si x es un número negativo.
    '''
    if x < 0:
        raise ValueError("x debe ser un entero no negativo")
    if x == 0 or x == 1:
        return 1
    return x * my_factorial(x - 1)

# Probar la función factorial para varios valores
print("Prueba de la función my_factorial:")
for num in range(1, 10, 2):
    print(f" - Factorial de {num}!: {my_factorial(num)}")
````

````python
# 2. Implementación de la Clase Persona para Personal del CSIC
class Persona:
    '''
    Clase que representa a un miembro del personal del CSIC.
    '''
    # Listas estáticas de validación de institutos y servicios del CSIC
    INSTITUTOS_CSIC = [
        "IMSE",         # Instituto de Microelectrónica de Sevilla (IMSE-CNM)
        "IMB",          # Instituto de Microelectrónica de Barcelona
        "EBD",          # Estación Biológica de Doñana
        "CSIC-URICI",   # Unidad de Recursos de Información Científica para la Investigación
        "CSIC-SGAI"     # Secretaría General Adjunta de Informática 
    ]

    SERVICIOS_CSIC = [
        "Biblioteca",
        "Informática",
        "Administración",
        "Comunicación",
        "Heladeria",
        "Investigación"
    ]

    def __init__(self, nombre: str, edad: int, instituto: str = None, servicio: str = None):
        self.nombre = nombre
        self.edad = edad
        self._instituto = None
        self._servicio = None
        
        # Asignar validando a través de los setters de propiedades
        if instituto:
            self.instituto = instituto
        if servicio:
            self.servicio = servicio

    # --- Propiedad INSTITUTO (Getter y Setter con validación) ---
    @property
    def instituto(self):
        return self._instituto

    @instituto.setter
    def instituto(self, valor):
        if valor is not None and valor not in self.INSTITUTOS_CSIC:
            raise ValueError(f"Instituto no válido: '{valor}'. Opciones válidas: {self.INSTITUTOS_CSIC}")
        self._instituto = valor

    # --- Propiedad SERVICIO (Getter y Setter con validación) ---
    @property
    def servicio(self):
        return self._servicio

    @servicio.setter
    def servicio(self, valor):
        if valor is not None and valor not in self.SERVICIOS_CSIC:
            raise ValueError(f"Servicio no válido: '{valor}'. Opciones válidas: {self.SERVICIOS_CSIC}")
        self._servicio = valor

    # --- Métodos de Comportamiento ---
    def saludar(self) -> str:
        '''Devuelve un saludo formal indicando la adscripción institucional.'''
        if self.instituto:
            return f"Hola, me llamo {self.nombre}, tengo {self.edad} años y pertenezco al instituto {self.instituto}."
        elif self.servicio:
            return f"Hola, me llamo {self.nombre}, tengo {self.edad} años y trabajo en el servicio de {self.servicio}."
        else:
            return f"Hola, me llamo {self.nombre}, tengo {self.edad} años."

    def crecer(self) -> int:
        '''Incrementa la edad de la persona en un año (cumpleaños).'''
        self.edad += 1
        return self.edad
````
