## 6. Desarrollo de Módulos: Funciones Creadas y Clases de Dominio Científico

A continuación implementamos en el código fuente de nuestro paquete dos componentes fundamentales:
1. Una función matemática recursiva (`my_factorial`).
2. Una clase de dominio para gestionar investigadores y personal de apoyo del CSIC (`Persona`).


````python
"""
Módulo usuario_lib.py
=====================
Librería de funciones matemáticas y utilidades para el paquete de Ciencia Abierta.
Autor: Curso Python para Ciencia Abierta (CSIC)
"""
def my_factorial(x: int) -> int:
    """
    Calcula el factorial de un número entero no negativo utilizando recursión.
    Parameters
    ----------
    x : int
        Número entero no negativo.
    Returns
    -------
    int
        El factorial de x.
    Raises
    ------
    ValueError
        Si x es un número negativo.
    """
    if x < 0:
        raise ValueError("x debe ser un entero no negativo")
    if x == 0:
        return 1
    if x == 1:
        return 1
    return x * my_factorial(x - 1)
# --- Bloque de Ejecución Directa para Testeo ---
if __name__ == "__main__":
    print("=" * 60)
    print("🧪 Probando el módulo usuario_lib.py localmente:")
    print("=" * 60)
    print("\nCálculo de factoriales de números impares del 1 al 9:")
    for num in range(1, 10, 2):
        print(f" - Factorial de {num}!: {my_factorial(num)}")
    print("\n✅ Pruebas locales completadas exitosamente.")
````

````python
"""
Módulo usuario_class.py
======================
Definición de la clase Persona para la gestión de adscripciones del personal del CSIC.

Autor: Curso Python para Ciencia Abierta (CSIC)
"""


class Persona:
    """
    Clase que representa a un miembro del personal del CSIC.
    """
    # Listas estáticas de validación de institutos y servicios del CSIC
    INSTITUTOS_CSIC = [
        "IMSE",         # Instituto de Microelectrónica de Sevilla (IMSE-CNM)
        "IMB",          # Instituto de Microelectrónica de Barcelona
        "EBD",          # Estación Biológica de Doñana
        "CSIC-URICI",   # Unidad de Recursos de Información Científica para la Investigación
        "CSIC-SGAI"     # Secretaría General Adjunta de Informática 
    ]

    SERVICIOS_CSIC = [
        "Biblioteca",
        "Informática",
        "Administración",
        "Comunicación",
        "Heladeria",
        "Investigación"
    ]

    def __init__(self, nombre: str, edad: int, instituto: str = None, servicio: str = None):
        self.nombre = nombre
        self.edad = edad
        self._instituto = None
        self._servicio = None
        
        # Asignar validando a través de los setters de propiedades
        if instituto:
            self.instituto = instituto
        if servicio:
            self.servicio = servicio

    # --- Propiedad INSTITUTO (Getter y Setter con validación) ---
    @property
    def instituto(self):
        return self._instituto

    @instituto.setter
    def instituto(self, valor):
        if valor is not None and valor not in self.INSTITUTOS_CSIC:
            raise ValueError(f"Instituto no válido: '{valor}'. Opciones válidas: {self.INSTITUTOS_CSIC}")
        self._instituto = valor

    # --- Propiedad SERVICIO (Getter y Setter con validación) ---
    @property
    def servicio(self):
        return self._servicio

    @servicio.setter
    def servicio(self, valor):
        if valor is not None and valor not in self.SERVICIOS_CSIC:
            raise ValueError(f"Servicio no válido: '{valor}'. Opciones válidas: {self.SERVICIOS_CSIC}")
        self._servicio = valor

    # --- Métodos de Comportamiento ---
    def saludar(self) -> str:
        """Devuelve un saludo formal indicando la adscripción institucional."""
        if self.instituto:
            return f"Hola, me llamo {self.nombre}, tengo {self.edad} años y pertenezco al instituto {self.instituto}."
        elif self.servicio:
            return f"Hola, me llamo {self.nombre}, tengo {self.edad} años y trabajo en el servicio de {self.servicio}."
        else:
            return f"Hola, me llamo {self.nombre}, tengo {self.edad} años."

    def crecer(self) -> int:
        """Incrementa la edad de la persona en un año (cumpleaños)."""
        self.edad += 1
        return self.edad


# --- Bloque de Ejecución Directa para Testeo ---
if __name__ == "__main__":
    print("=" * 60)
    print("🧪 Probando el módulo usuario_class.py localmente:")
    print("=" * 60)

    # Instanciar miembro del CSIC
    persona_test = Persona(nombre="Gustavo Liñán", edad=53, instituto="IMSE")
    print(f"\n1. Saludo inicial: {persona_test.saludar()}")

    # Probar método crecer
    persona_test.crecer()
    print(f"2. Cumpleaños: {persona_test.saludar()}")

    # Probar cambio de adscripción a un servicio
    persona_test.instituto = None
    persona_test.servicio = "Heladeria"
    print(f"3. Cambio de servicio: {persona_test.saludar()}")

    print("\n✅ Pruebas locales de la clase Persona completadas con éxito.")
````

## 7.  La Importancia de las Pruebas Unitarias (*Testing*) en Ciencia Abierta

En el desarrollo de software científico y la investigación abierta, escribir **pruebas unitarias** (*unit tests*) es la garantía fundamental de que **nuestros cálculos, algoritmos y flujos de datos son correctos, estables y reproducibles**.

#### 🎯 ¿Por qué es vital definir tests para las funciones que creamos?

1. **Garantía de Correctitud Científica:**
   Un fallo no detectado en una función matemática o en una rutina de procesamiento puede alterar los resultados de un experimento. Las pruebas verifican de forma automatizada que las funciones devuelven exactamente los valores esperados ante casos límite y de uso normal.

2. **Prevención de Regresiones (Evitar romper lo que funcionaba):**
   A medida que el paquete crece o refactorizamos código, ejecutar las pruebas nos avisa de inmediato si un nuevo cambio ha roto una funcionalidad que previamente funcionaba bien.

3. **Confianza en el Trabajo Colaborativo:**
   En proyectos con varios investigadores o grupos de trabajo del CSIC, los tests permiten que cualquier miembro añada mejoras sin miedo a desestabilizar el trabajo de sus compañeros.

4. **Documentación Viva y Verificable:**
   Un archivo de prueba (`test_*.py`) es el mejor ejemplo práctico de uso: muestra de manera exacta e inequívoca qué entradas recibe una función, qué salidas debe generar y qué excepciones debe lanzar.

5. **Integración Continua (CI/CD con GitHub Actions):**
   En paquetes modernos (como el generado por **pyOpenSci**), las pruebas se ejecutan automáticamente en servidores remotos cada vez que enviamos cambios a GitHub, garantizando que el paquete funcione en diferentes sistemas operativos y versiones de Python.

## NOTA: El Gestor de Contexto `with pytest.raises(...)` en las pruebas ##


### En el desarrollo de software de calidad, **no solo debemos probar que las funciones funcionen bien cuando se les proporcionan datos correctos** (los llamados *"casos felices"*); también es imprescindible verificar que el programa **reaccione correctamente e interrumpa la ejecución controladamente** cuando recibe datos inválidos o erróneos.
---
#### ❓ ¿Qué hace la instrucción `with` con `pytest.raises`?
La palabra clave **`with`** abre lo que en Python se denomina un **Gestor de Contexto** (*Context Manager*). Al combinarlo con `pytest.raises(...)`:
1. **Captura el Error:** Le dice a `pytest`: *"Ejecuta el bloque de código indentado a continuación y espera que se produzca una excepción del tipo especificado"*.
2. **Pasa el test si se lanza el error:** Si la línea dentro del bloque efectivamente lanza la excepción esperada (por ejemplo, `ValueError`), **el test se considera EXITOSO (PASSED)**.
3. **Falla el test si NO hay error:** Si el código dentro del bloque se ejecuta limpiamente sin lanzar ningún error, **el test FALLA (FAILED)** porque la validación de seguridad de nuestro código no se activó.
