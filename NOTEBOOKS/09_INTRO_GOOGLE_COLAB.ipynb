## 2. ¿Qué es Google Colab? Ventajas, Inconvenientes y Comparativa

**Google Colaboratory (Colab)** es un servicio de cuaderno de Jupyter alojado en la nube por Google. Permite escribir, ejecutar y compartir código Python directamente en el navegador web sin necesidad de instalar o configurar ningún software localmente.

Cada cuaderno de Colab se ejecuta en una **máquina virtual Linux dedicada** (con recursos de CPU, RAM y espacio en disco) provista dinámicamente por los servidores de Google.

---

### 🟢 Ventajas Principales para la Ciencia Abierta:

1. **Configuración Cero:** No requiere instalar Python, entornos virtuales ni librerías locales. Con solo una cuenta de Google puedes empezar a codificar.
2. **Acceso Gratuito a Hardware de Alto Rendimiento (GPU/TPU):** Proporciona acceso a acelardores de hardware (GPU NVIDIA T4/L4 o TPU) esenciales para *Machine Learning* y simulaciones pesadas.
3. **Ecosistema Científico Preinstalado:** Incluye por defecto las principales librerías de datos (`pandas`, `numpy`, `scipy`, `scikit-learn`, `matplotlib`, `seaborn`, `tensorflow`, `torch`).
4. **Colaboración en Tiempo Real:** Permite que varios investigadores trabajen, comenten y editen el mismo cuaderno simultáneamente, con control de avatares visibles en la interfaz.
5. **Integración con Asistentes de IA (Gemini):** Incorpora ayuda en línea con Gemini para autocompletar código, explicar errores y sugerir refactorizaciones.
6. **Persistencia e Integración:** Guardado automático en **Google Drive** y exportación directa a repositorios de **GitHub**.

---

### 🔴 Inconvenientes y Limitaciones a Tener en Cuenta:

1. **Sesiones Efímeras (Tiempo de Ejecución):** Las instancias en la nube son temporales. Si el cuaderno permanece inactivo o supera el tiempo máximo por sesión (generalmente 12 horas en la versión gratuita), la máquina se desconecta y los archivos locales del directorio `/content/` se borran.
2. **Dependencia de Conexión a Internet:** Requiere conexión continua y estable a la red.
3. **Políticas de Privacidad y Confidencialidad:** Al subir datos a Colab, estos residen temporalmente en la infraestructura de Google. Se debe verificar el nivel de sensibilidad y protección de datos antes de subir proyectos confidenciales del CSIC.

## 4. Análisis Exploratorio de Datos (EDA) en la Nube: Cáncer de Tiroides

Analizaremos un conjunto de datos clínico real obtenido de Kaggle/UCI sobre la **recurrencia del cáncer de tiroides bien diferenciado**. Contiene 13 variables clínico-patológicas recopiladas durante 15 años con un seguimiento mínimo de 10 años por paciente.

### Atributos Clave:
- **`Age` / `Gender` / `Smoking`:** Edad, género e historial de tabaquismo.
- **`Thyroid Function`:** Evaluación clínica de la función tiroidea.
- **`Physical Examination`:** Diagnóstico físico de bocio o nódulos.
- **`Adenopathy`:** Presencia o ausencia de adenopatías linfáticas en el cuello.
- **`Pathology`:** Histología tumoral (Papilar, Folicular, Células de Hurthle, Micro-papilar).
- **`Risk`:** Clasificación de riesgo de recurrencia (Bajo, Intermedio, Alto).
- **`T`, `N`, `M`:** Sistema de clasificación TNM de tumores sólidos.
- **`Response`:** Respuesta al tratamiento (Excelente, Indeterminada, Incompleta).
- **`Recurred`:** Recurrencia de la enfermedad (Sí / No).


```python
# Importación de librerías para análisis y visualización
import sys
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración de estilo gráfico limpio
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'

df = pd.read_csv(target_csv)

print(f"Forma del DataFrame: {df.shape[0]} filas, {df.shape[1]} columnas")
print("Primeras 5 filas del dataset original:")
print(df.head(5))
```

```python
# Diagnóstico de valores nulos o ausentes
nas_totales = df.isna().sum().sum()
nulls_totales = df.isnull().sum().sum()

print("=" * 60)
print(f"Recuento total de valores NA: {nas_totales}")
print(f"Recuento total de valores NULL: {nulls_totales}")

if nas_totales == 0 and nulls_totales == 0:
    print("✅ El conjunto de datos está 100% completo (sin valores faltantes).")
else:
    print("⚠️ Se requieren tareas de imputación o limpieza de nulos.")
print("=" * 60)
```

## 5. Curación de Metadatos: Traducción de Columnas y Categorías

Para facilitar la interpretación científica del equipo clínico en España, traduciremos y normalizaremos los nombres de las columnas y los valores de las variables categóricas al español.


```python
# Renombrar columnas a términos científicos en español
columnas_es = {
    'Age': 'Edad',
    'Gender': 'Género',
    'Smoking': 'Fumador',
    'Hx Smoking': 'H. tabaquismo',
    'Hx Radiothreapy': 'H. radioterapia',
    'Thyroid Function': 'Función tiroidea',
    'Physical Examination': 'Examen físico',
    'Adenopathy': 'Adenopatía',
    'Pathology': 'Patología',
    'Focality': 'Focalidad',
    'Risk': 'Riesgo',
    'T': 'Tumor',
    'N': 'Ganglio Linfático',
    'M': 'Metastasis',
    'Stage': 'Estadio',
    'Response': 'Respuesta',
    'Recurred': 'Recurrente'
}

df.rename(columns=columnas_es, inplace=True)
print("Columnas renombradas exitosamente:")
print(list(df.columns))
```

```python
# Normalización y traducción de valores categóricos clave
df['Adenopatía'] = df['Adenopatía'].replace({
    'No': 'Sin Adenopatía',
    'Left': 'Lado Izquierdo',
    'Right': 'Lado Derecho',
    'Extensive': 'Extensa y Generalizada'
})

df['Función tiroidea'] = df['Función tiroidea'].replace({
    'Normal': 'Normal',
    'Euthyroid': 'Eutiroidismo',
    'Clinical Hyperthyroidism': 'Hipertiroidismo Clínico',
    'Clinical Hypothyroidism': 'Hipotiroidismo Clínico',
    'Subclinical Hyperthyroidism': 'Hipertiroidismo Subclínico',
    'Subclinical Hypothyroidism': 'Hipotiroidismo Subclínico'
})

df['Examen físico'] = df['Examen físico'].replace({
   'Single nodular goiter-left': 'Bocio nodular único - izq.',
   'Multinodular goiter': 'Bocio multinodular',
   'Single nodular goiter-right': 'Bocio nodular único - der.',
   'Diffuse goiter': 'Bocio difuso',
   'Normal': 'Normal'
})

df['Patología'] = df['Patología'].replace({
    'Micropapillary': 'Micro-papilar',
    'Papillary': 'Papilar',
    'Follicular': 'Folicular',
    'Hurthel cell': 'Células de Hurthle'
})

df['Riesgo'] = df['Riesgo'].replace({
    'Low': 'Bajo',
    'Intermediate': 'Intermedio',
    'High': 'Alto'
})

df['Respuesta'] = df['Respuesta'].replace({
    'Indeterminate': 'Indeterminada',
    'Excellent': 'Excelente',
    'Structural Incomplete': 'Incompleta Estructural',
    'Biochemical Incomplete': 'Incompleta Bioquímica'
})

# Estandarizar respuestas binarias Yes/No a s/n
df.replace({'Yes': 's', 'No': 'n'}, inplace=True)

print("Categorías traducidas correctamente. Muestra de 'Respuesta':")
print(df['Respuesta'].value_counts())
```

```python
# Inserción de un nuevo registro de ensayo clínico (Simulación)
paciente_control = {
    'Edad': 47,
    'Género': 'M',
    'Fumador': 'n',
    'H. tabaquismo': 'n',
    'H. radioterapia': 'n',
    'Función tiroidea': 'Eutiroidismo',
    'Examen físico': 'Bocio nodular único - der.',
    'Adenopatía': 'Sin Adenopatía',
    'Patología': 'Papilar',
    'Focalidad': 'Uni-Focal',
    'Riesgo': 'Bajo',
    'Tumor': 'T2a',
    'Ganglio Linfático': 'N0',
    'Metastasis': 'M0',
    'Estadio': 'II',
    'Respuesta': 'Excelente',
    'Recurrente': 'n'
}

df = pd.concat([df, pd.DataFrame([paciente_control])], ignore_index=True)

# Exportar dataset curado
output_curated_path = os.path.join(datasets_dir, "Thyroid_Diff_ES.csv")
df.to_csv(output_curated_path, index=False)

print(f"Nuevo registro concatenado. Dataset guardado en: '{output_curated_path}'")
```

## 6. Visualización de Patrones Clínicos: Seaborn y Matplotlib en Colab

Generaremos figuras compuestas para analizar la relación entre la edad de los pacientes, el estadio del tumor, la categoría de riesgo y la respuesta al tratamiento.


```python
# 1. Diagramas de Caja (Boxplots): Distribución de Edad por Estadio del Tumor y del Cáncer
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Boxplot por Tumor
sns.boxplot(x='Tumor', y='Edad', data=df, palette="OrRd", hue='Tumor', legend=False, ax=axes[0])
axes[0].set_title('Distribución de Edad por Clasificación de Tumor (T)', fontweight='bold')
axes[0].set_xlabel('Estadio del Tumor')
axes[0].set_ylabel('Edad del Paciente')
axes[0].tick_params(axis='x', rotation=45)

# Boxplot por Estadio General
sns.boxplot(x='Estadio', y='Edad', data=df, palette="PuBu", hue='Estadio', legend=False, ax=axes[1])
axes[1].set_title('Distribución de Edad por Estadio Clínico General', fontweight='bold')
axes[1].set_xlabel('Estadio General')
axes[1].set_ylabel('Edad del Paciente')

plt.tight_layout()
plt.show()
```

```python
# 2. Histogramas de Distribución por Edad y Género
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histograma de Edad con Estimación de Densidad de Kernel (KDE)
sns.histplot(df['Edad'], bins=20, kde=True, color='steelblue', ax=axes[0])
axes[0].set_title('Distribución Etaria de la Cohorte', fontweight='bold')
axes[0].set_xlabel('Edad')
axes[0].set_ylabel('Frecuencia (Nº Pacientes)')

# Conteo por Género
sns.countplot(x='Género', data=df, palette=['pink', 'lightblue'], ax=axes[1])
axes[1].set_title('Distribución por Género', fontweight='bold')
axes[1].set_xlabel('Género (F: Femenino, M: Masculino)')
axes[1].set_ylabel('Frecuencia')

plt.tight_layout()
plt.show()
```

```python
# 3. Gráficos de Violín (Violin Plots): Edad por Categoría de Riesgo
fig, ax = plt.subplots(figsize=(9, 5))

sns.violinplot(x='Riesgo', y='Edad', data=df, inner='quartile', palette='Set2', ax=ax)
ax.set_title('Distribución de Edades por Categoría de Riesgo Clínico', fontsize=12, fontweight='bold')
ax.set_xlabel('Categoría de Riesgo')
ax.set_ylabel('Edad')

plt.show()
```

## 3. Dashboards Dinámicos con `ipywidgets` y `@interact`

`ipywidgets` es el estándar abierto de la comunidad Jupyter. El decorador `@interact` inspecciona los argumentos de una función de Python y crea automáticamente los controles adecuados.


```python
# Carga e Ingesta del Dataset Curado (Thyroid_Diff_ES.csv)
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from ipywidgets import interact, IntSlider, Dropdown, ToggleButtons

# Estilo gráfico limpio
sns.set_theme(style="whitegrid", palette="muted")

# Determinar ruta flexible del dataset
datasets_dir = "../DATASETS" if os.path.exists("../DATASETS") else "DATASETS"
csv_path = os.path.join(datasets_dir, "Thyroid_Diff_ES.csv")

if not os.path.exists(csv_path):
    csv_path = os.path.join(datasets_dir, "Thyroid_Diff.csv")

df_widget = pd.read_csv(csv_path if os.path.exists(csv_path) else "Thyroid_Diff_ES.csv")

print(f"Dataset cargado para exploración interactiva. Registros totales: {len(df_widget)}")
```

```python
# Dashboard Clínico Interactivo en Tiempo Real
@interact(
    edad_limite=IntSlider(min=25, max=80, step=5, value=55, description='Edad Máx:'),
    filtro_riesgo=Dropdown(options=['Todos', 'Bajo', 'Intermedio', 'Alto'], value='Todos', description='Riesgo:'),
    vista_grafico=ToggleButtons(options=['Boxplot', 'Violinplot', 'Histograma'], description='Visualización:')
)
def dashboard_clinico_interactivo(edad_limite, filtro_riesgo, vista_grafico):
    # 1. Filtrar DataFrame dinámicamente
    df_sub = df_widget[df_widget['Edad'] <= edad_limite].copy()
    if filtro_riesgo != 'Todos' and 'Riesgo' in df_sub.columns:
        df_sub = df_sub[df_sub['Riesgo'] == filtro_riesgo]

    print(f"📊 Muestra filtrada: {len(df_sub)} pacientes (Edad <= {edad_limite}, Riesgo = '{filtro_riesgo}')")

    # 2. Renderizar gráfico dinámico
    fig, ax = plt.subplots(figsize=(10, 4.5))

    if vista_grafico == 'Boxplot':
        col_x = 'Estadio' if 'Estadio' in df_sub.columns else 'Stage'
        sns.boxplot(x=col_x, y='Edad', data=df_sub, palette='Set2', hue=col_x, legend=False, ax=ax)
        ax.set_title(f'Distribución de Edad por Estadio Clínico (Riesgo: {filtro_riesgo})', fontweight='bold')
    elif vista_grafico == 'Violinplot':
        col_x = 'Respuesta' if 'Respuesta' in df_sub.columns else 'Response'
        sns.violinplot(x=col_x, y='Edad', data=df_sub, palette='Pastel1', ax=ax)
        ax.set_title(f'Edad por Respuesta al Tratamiento (Riesgo: {filtro_riesgo})', fontweight='bold')
        ax.tick_params(axis='x', rotation=20)
    else:
        sns.histplot(df_sub['Edad'], bins=15, kde=True, color='teal', ax=ax)
        ax.set_title(f'Histograma (N = {len(df_sub)})', fontweight='bold')
        ax.set_xlabel('Edad (años)')
        ax.set_ylabel('Nº de Pacientes')

    plt.tight_layout()
    plt.show()
```

## 4. Visualizador Multivariable con Botón de Ejecución (`interact_manual`)

Cuando el análisis o la generación de figuras es computacionalmente intensivo, es recomendable utilizar `interact_manual`. Esto añade un botón **"Run Interact"** para que el gráfico solo se vuelva a calcular cuando el usuario confirme los cambios.


```python
# Generador de Figuras Bajo Demanda (interact_manual)
from ipywidgets import interact_manual

@interact_manual(
    variable_x=Dropdown(options=['Estadio', 'Tumor', 'Riesgo', 'Patología'], value='Estadio', description='Variable X:'),
    paleta_color=Dropdown(options=['viridis', 'magma', 'Set2', 'PuBu', 'Spectral'], value='Set2', description='Paleta:'),
    mostrar_puntos=True
)
def reporte_multivariable_manual(variable_x, paleta_color, mostrar_puntos):
    if variable_x not in df_widget.columns:
        print(f"La columna '{variable_x}' no se encuentra en el DataFrame.")
        return

    fig, ax = plt.subplots(figsize=(10, 5))

    # Boxplot base
    sns.boxplot(x=variable_x, y='Edad', data=df_widget, palette=paleta_color, hue=variable_x, legend=False, ax=ax)

    # Superponer puntos individuales (Stripplot) si está marcado
    if mostrar_puntos:
        sns.stripplot(x=variable_x, y='Edad', data=df_widget, color='black', alpha=0.3, jitter=0.2, size=5, ax=ax)

    ax.set_title(f'Análisis Exploratorio: Edad vs. {variable_x}', fontsize=13, fontweight='bold')
    ax.set_xlabel(variable_x)
    ax.set_ylabel('Edad del Paciente')
    ax.tick_params(axis='x', rotation=30)

    plt.tight_layout()
    plt.show()
```
