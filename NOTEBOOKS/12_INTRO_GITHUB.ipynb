## 2. Control de Versiones en la Nube, Privacidad y Licencias Abiertas

Publicar nuestro código en repositorios remotos como **GitHub** es una pieza angular de la **Ciencia Abierta**. Facilita que otros grupos de investigación auditen, reproduzcan y reutilicen nuestros métodos computacionales.

### 🔒 Repositorios Públicos vs. Privados
- **Repositorio Privado:** El código solo es accesible por ti y por los colaboradores explícitamente invitados. Es ideal para las fases iniciales de un proyecto, o casi obligatorio antes de la publicación oficial del artículo científico.
- **Repositorio Público:** El código es visible para toda la comunidad internacional. Fomenta las citas científicas, la transparencia y la colaboración.

> **💡 Consejo Pedagógico:** Puedes comenzar tu investigación en un repositorio privado en GitHub y, una vez aceptado el manuscrito en la revista científica, cambiar su visibilidad a **Público** con un solo clic en la llamada **zona de peligro** de GitHub.

---

### 🛡️ Protección de Secretos y Datos Sensibles (`.gitignore`)
Nunca debemos subir al repositorio remoto:
1. **Archivos de entorno y credenciales (`.env`):** Claves de API, contraseñas de bases de datos o tokens de acceso a repositorios institucionales.
2. **Datos con restricciones de privacidad:** Datos personales de pacientes, muestras sensibles no anonimizadas, conjuntos de datos masivos, datos sometidos a NDA, etc...MUCHO CUIDADO!!!!!
3. **Archivos generados automáticamente:** Entornos virtuales (`venv/`), carpetas de caché (`__pycache__/`) o binarios de compilación.

Para ello, verificamos que nuestro archivo **`.gitignore`** incluya estas reglas antes del primer commit.

---

### 📜 Licencias de Código Abierto
Un repositorio público sin licencia se considera con *"todos los derechos reservados"*, lo que impide legalmente que otros científicos reutilicen tu software. Las licencias más comunes en Ciencia Abierta son:
- **MIT License:** Permisiva y muy sencilla. Permite a cualquiera reutilizar, modificar y distribuir tu código libremente, siempre que mantengan la nota de copyright.
- **Apache License 2.0:** Permisiva, pero incluye una concesión explícita de derechos de patente.
- **GNU GPLv3:** Licencia *Copyleft*. Exige que cualquier obra derivada sea publicada también bajo la misma licencia libre GPL.

Prefijos standard en los **commits**

- refactor:	Reorganizar código sin cambiar su comportamiento.	refactor: simplificar bucle de filtrado en Pandas

## 5. Paso 3: Flujo de Trabajo con Ramas (*Feature Branch Workflow*)

### 🌿 ¿Qué es una Rama (*Branch*) en Git?

Una **rama** en Git es simplemente una **línea de desarrollo independiente y paralela**. Imagina que tu repositorio es un árbol:

- **La rama principal (`main`):** Representa el tronco del árbol. Es la versión oficial, estable, probada y lista para producción o publicación de tu proyecto.
- **Una rama secundaria (`feature/...`):** Es una ramificación que nace a partir de `main` y crea una copia exacta de tu código en un espacio de trabajo aislado.

Cualquier cambio, experimento o error que introduzcas en una rama secundaria **no afectará para nada al código de `main`** hasta que decidas fusionarlo explícitamente.

---

### 🎯 ¿Por qué debemos trabajar con ramas en Ciencia Abierta?

1. **Protección del Código Estable:** Mantiene la rama `main` siempre limpia y en perfecto estado funcional.
2. **Experimentación Sin Riesgos:** Te permite desarrollar nuevas funciones o probar ideas arriesgadas sin miedo a estropear el trabajo de tus compañeros.
3. **Revisión por Pares (*Code Review*):** Facilita que otros investigadores del equipo revisen tus cambios antes de incorporarlos definitivamente al proyecto.

## 6. Paso 4: Pull Requests (PR) y Revisión de Código en GitHub

Un **Pull Request (PR)** es una solicitud formal para fusionar el código de nuestra rama secundaria (`feature/nuevos-tests-persona`) en la rama principal (`main`).

### 🔄 Flujo de Trabajo de un Pull Request:
1. **Apertura del PR:** Al hacer `git push` de tu rama, entra a la página de tu repositorio en GitHub (`https://github.com/<user_name>/<user_name>_PYTHON_CIENCIA_ABIERTA_2`).
2. Haz clic en el botón destacado **"Compare & pull request"**.
3. **Descripción:** Escribe un título descriptivo y explica qué cambios introduce tu PR.
4. **Revisión de Código (*Diffs*):** Los colaboradores o revisores del grupo pueden examinar línea por línea los cambios realizados (líneas verdes `+` agregadas y líneas rojas `-` eliminadas).
5. **Aprobación y Fusión (*Merge*):** Una vez aprobado, se pulsa el botón **"Merge pull request"** y los cambios pasan a formar parte oficial de la rama `main`.
6. **Sincronización Local:** En tu máquina local, regresas a `main` y descargas las novedades:
   ```bash
   git checkout main
   git pull origin main
   ```


## 7. Paso 5: Integración Continua (CI/CD) con GitHub Actions

La **Integración Continua (CI)** consiste en automatizar la compilación, verificación y ejecución de pruebas unitarias de nuestro proyecto cada vez que alguien envía nuevo código a GitHub.

### 🤖 ¿Cómo funciona GitHub Actions en nuestro Paquete?

Cuando generamos la plantilla con **pyOpenSci**, `copier` creó automáticamente una carpeta oculta `.github/workflows/` con el archivo **`test.yml`**:

```text
PYTHON_CIENCIA_ABIERTA_2/
└── .github/
    └── workflows/
        └── test.yml        <-- Configuración del flujo de automatización de Pytest
```

### ⚙️ ¿Qué hace el archivo `test.yml` en la Nube?

Cada vez que haces `git push` o abres un Pull Request:
1. **GitHub arranca una máquina virtual remota (Ubuntu Linux) en la nube.**
2. **Instala Python (por ejemplo, versión 3.10, 3.11, 3.12).**
3. **Instala las dependencias de tu proyecto (`pyproject.toml`).**
4. **Ejecuta automáticamente la orden `pytest`.**
5. Si todos los tests pasan, GitHub asigna una marca verde 🟢 **(PASSED)** a tu commit. Si algún test falla, muestra una cruz roja 🔴 **(FAILED)** y envía una alerta por correo electrónico.

### ⚙️ Vamos a editar el archivo `test.yml`. Para ello:

1. **Navega al directorio `.github/workflows/test.yml`del repositorio**
2. **Edita el fichero y busca:**
    strategy:
        fail-fast: false
        matrix:
            os: [ubuntu-latest, macos-latest, windows-latest]
            python-version: ["3.10", "3.13"]
3. **Haz que el proyecto se testee para las versiones de python 3.11,3,12 y 3.13 cambiando la línea python-version asi:**
            python-version: ["3.11", "3.12", "3.13"]

### Ejercicio Práctico 2: Auditoría de Seguridad de Secretos
Comprueba el estado del archivo `.gitignore` de tu repositorio y confirma que la regla `.env` esté presente para evitar subir claves privadas a la nube.


````python
# Solución Interactiva Ejercicio 2: Verificación de Reglas de Ignorado en Python
import os

gitignore_path = os.path.join("..", ".gitignore")

if os.path.exists(gitignore_path):
    with open(gitignore_path, "r", encoding="utf-8") as f:
        reglas = [line.strip() for line in f if line.strip() and not line.startswith("#")]
    
    print("🔒 Reglas activas detectadas en .gitignore:")
    for r in reglas:
        print(f"  - {r}")
    
    print("")
    if any(".env" in r for r in reglas):
        print("✅ ¡Seguridad Confirmada! La regla .env está activa para proteger secretos.")
    else:
        print("⚠️ ALERTA: No se detectó la regla .env en .gitignore.")
else:
    print(f"ℹ️ El archivo .gitignore no se encuentra en la ruta esperada: {gitignore_path}")
````
