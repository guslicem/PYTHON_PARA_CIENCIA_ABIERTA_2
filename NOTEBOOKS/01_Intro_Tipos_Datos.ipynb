### 3. Uso de `print()` con formateo de f-strings

Mi consejo es que siempre imprimas con el `print(f"...")` o `print(f' ')`.

El operador `f"..."` (Python 3.6+, PEP 498), también llamado formateo de cadena mediante f-strings, te permite insertar variables o expresiones directamente dentro de una cadena de texto.


````python
valor = 3.141592653589793238462643383279502
print(f'El valor de pi es {valor}')
print(f"El valor de pi con 1 decimal   {valor:.1f}")  
print(f"El valor de pi con 2 decimales {valor:.2f}")  
print(f"El valor de pi con 4 decimales {valor:.4f}") 
print(f"El valor de pi con 6 decimales {valor:.6f}")  
````

````python
Seconds_In_A_Year = 60 * 60 * 24 * 365
MiEdad = 53

print(f"He vivido más de {MiEdad * Seconds_In_A_Year:.1e} segundos (Formato exponencial con 1 decimal)") 
print(f"He vivido más de {MiEdad * Seconds_In_A_Year:.3e} segundos (Formato exponencial con 3 decimales)") 
print(f"He vivido más de {MiEdad * Seconds_In_A_Year:>30} segundos (Valor entero exacto con espaciado)") 
````

````python
nombre = "Ana"
edad = 27
print(f"{nombre} tiene {edad} años.")  # Ana tiene 27 años

# También puedes realizar operaciones directamente dentro del f-string
print(f"El doble de pi es {2 * valor:.3f}")  # El doble de pi es 6.283

print("Demostración de alineación dinámica con f-strings:")
for sep in range(2, 20, 4):
    print(f"{nombre:>{sep}}, {edad:>{sep}}")
````

````python
# Una de las ventajas de Python es que no es necesario declarar tipos... se infieren en tiempo de ejecución

# Tipos numéricos y operaciones
x1 = 10         # Un entero (int)
r1 = 3.14159    # Un float (real)
z1 = 2 + 3j     # Un complejo (complex)

print(f"\tEntero: {x1:>10}   -- tipo: {type(x1)}")       # \t mete una tabulación
print(f"\tFlotante: {r1:>10} -- tipo: {type(r1)}")
print(f"\tComplejo: {z1:>10} -- tipo: {type(z1)}")

z1 = 4          # Cambio el tipo de una variable ya declarada. Tipado Dinámico
print(f"\tAhora z1  {z1:>10} -- tipo: {type(z1)}")
````

## 4. Operador Módulo `%` en Python

El operador **`%`** devuelve el resto de una división.
Es muy útil para comprobar divisibilidad, determinar si un número es par o impar, o realizar operaciones cíclicas (por ejemplo, rotar índices en una lista).

---

### Sintaxis Básica
```python
resto = a % b
```


````python
# División simple con resto
a = 10
b = 3

resto = a % b
print(f"10 % 3 = {resto} (salida: 1, porque 10 ÷ 3 = 3 y sobra 1)")

print(f"15 % 6 = {15 % 6} (salida: 3, porque 15 ÷ 6 = 2 y sobra 3)")

# Comprobar si un número es impar usando %
Es_Impar = (a % 2 == 1)
print("El número", a, "es impar?", Es_Impar, "tipo:", type(Es_Impar))
````

## 5. Operador División Entera `//` en Python

El operador **`//`** realiza una división entera, es decir, devuelve solo la parte entera del cociente, descartando los decimales.
Es muy útil cuando quieres saber cuántas veces un número cabe dentro de otro sin preocuparte del resto.


````python
a = 23
b = 7

cociente = a // b
print(a, "//", b, "es", cociente)

# Combinando con el operador módulo (%)
print(f"El resultado de {a}//{b} es {a//b} con {a%b} de resto") 
````

````python
# Operando con variables de distinto o mismo tipo
x2 = 4
r2 = 2.12**2   # Para elevar al cuadrado
z2 = 3 - 6j

x3 = x1 + x2
r3 = r1 + r2
z3 = z1 + z2

print(f"Suma enteros S = {x3}")
print(f"Suma enteros directamente en el print sin asignar a variable: S = {x1 + x2}")
print(f"Suma floats {r1:.3f} + {r2:.3f} restringida a 3 decimales: S = {r3:.3f}")
print(f"Suma complejos {z1} + {z2} = {z3}")
````

````python
# Sumamos un real y un entero
m1 = x1 + r1
m2 = x1 * z1

print(f"Suma de distintos tipos {x1} + {r1} = {m1} tipo: {type(m1)}")

# Convertimos m1 en entero de nuevo
m1 = int(m1)
print(f"Suma de distintos tipos convertida a entero {x1} + {r1:.3f} = {m1} tipo: {type(m1)}")

print(f"Entero x complejo {x1} x {z1} = {m2} tipo: {type(m2)}")
print(f"Real x complejo {r1:.3f} x {z1} = {r1*z1:.3f} tipo: {type(r1*z1)}")
print(f"Complejo x complejo {z1} x {z2} = {z1*z2} tipo: {type(z1*z2)}")
````

````python
# Tipado dinámico y conversión a string / bool
x = "una cadena"
print("x ahora es:", x, "y su tipo es", type(x))

a = 0.0014
b = str(a)
c = bool(a)

print(f"a = {a}, tipo: {type(a)}")
print(f"b = '{b}', tipo: {type(b)}")
print(f"c = {c}, tipo: {type(c)}")
````

# 6. Gestión Básica NATIVA de Cadenas en Python (sin librerías externas)

En Python, una cadena (`str`) es una secuencia inmutable de caracteres.

## Concatenación
`saludo = texto1 + ", " + texto2`

## Sustitución
`nuevo_saludo = saludo.replace("Mundo", "Python")`

## Comparación
`texto1 == texto2`

## Método `.join()`
Su trabajo es unir todos los elementos de una lista en una sola cadena usando el string llamante como separador:
`separador.join(lista)`

### Normalización de Títulos y Preservación de Siglas Científicas:
Al usar `.strip().title()`, siglas como `"CSIC"` pasarían a `"Csic"`. Para evitarlo, usamos la lista:
`words_to_ignore = ['CSIC', 'USE', 'NASA']`.


````python
# Cadenas, operaciones y join
texto1 = "Hola"
texto2 = "Mundo"

# ---> Dividir la celda!!!! 

# Concatenación
saludo = texto1 + ", " + texto2
print(saludo)

# Sustitución
nuevo_saludo = saludo.replace("Mundo", "Python")
print(f"{nuevo_saludo} ---> longitud: {len(nuevo_saludo)}")

# Comparación
print(f"¿Son iguales? {texto1 == texto2}")

# Uso de .join() con autores científicos
autores = ["Einstein, A.", "Curie, M.", "Feynman, R."]
autores_str = "; ".join(autores)
print(f"Autores unificados: {autores_str}")

# Tratamiento de siglas científicas (words_to_ignore)
raw_title = "   análisis exploratorio de datos de investigación CSIC   "
clean_title = raw_title.strip().title()

words_to_ignore = ["CSIC", "USE", "NASA"]
for word in words_to_ignore:
    clean_title = clean_title.replace(word.title(), word)

print("")
print(f"Título original: '{raw_title}'")
print(f"Título limpio (siglas preservadas): '{clean_title}'")
````

````python

# Normalización de DOI
doi_raw = "  DOI: 10.1038/s41586-020-2649-2 "
doi_clean = doi_raw.strip().replace("DOI:", "").strip()
print(f"DOI normalizado: https://doi.org/{doi_clean}")

# Intentemos cambiar un elemento... RECUERDA: LAS CADENAS SON INMUTABLES
try:
    nuevo_saludo[0] = 'C'  # Esto lanza un TypeError
except TypeError as e:
    print("")
    print(f"✔ Capturado TypeError esperado (Inmutabilidad de cadenas): {e}")
````

````python
# Forma avanzada de modificar cadenas mediante rebanado (Slicing):
texto = "hola"
i = 2  # Posición que queremos cambiar (la 'l')

nuevo_caracter = 'K'

if 0 <= i < len(texto):
    texto_modificado = texto[:i] + nuevo_caracter + texto[i+1:]
    print(f"Texto original:   {texto}\n\n")
    print(f"Texto modificado: {texto_modificado}")
else:
    print("Índice fuera de rango")
````
